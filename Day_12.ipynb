# 21 Days Python Challenge
## Day 12 🫪

```python
double = lambda number: number * 2
print(double(6))
```

```python
add = lambda a, b: a + b
print(add(5, 7))
```

```python
numbers = [1, 2, 3, 4]
print(list(map(lambda x: x ** 2, numbers)))
```

```python
numbers = [1, 2, 3, 4, 5]
print(list(filter(lambda x: x % 2, numbers)))
```

```python
names = ['Ravi', 'Asha', 'Mira']
print(sorted(names, key=lambda name: len(name)))
```

```python
squares = [number ** 2 for number in range(1, 6)]
print(squares)
```

```python
even_numbers = [number for number in range(10) if number % 2 == 0]
print(even_numbers)
```

```python
labels = ['even' if number % 2 == 0 else 'odd' for number in range(1, 6)]
print(labels)
```

```python
lengths = {word: len(word) for word in ['cat', 'python', 'book']}
print(lengths)
```

```python
pairs = [(x, y) for x in range(1, 3) for y in range(1, 3)]
print(pairs)
```
